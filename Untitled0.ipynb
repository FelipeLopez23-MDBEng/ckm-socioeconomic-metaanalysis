```r
pkgs <- c("haven", "dplyr", "tidyr", "survey", "metafor", "ggplot2", "patchwork")
new_pkgs <- pkgs[!(pkgs %in% installed.packages()[, "Package"])]
if (length(new_pkgs)) install.packages(new_pkgs)

library(haven)
library(dplyr)
library(tidyr)
library(survey)
library(metafor)
library(ggplot2)
library(patchwork)

options(survey.lonely.psu = "adjust")

load_cycle <- function(yr_dir, suffix, label, is_pre = FALSE) {
  base <- paste0("https://wwwn.cdc.gov/Nchs/Data/Nhanes/Public/", yr_dir, "/DataFiles/")
  mods <- c("DEMO", "BMX", "GHB", "GLU", "TCHOL", "HDL", "TRIGLY", "BIOPRO", "ALB_CR", "MCQ", "BPQ", "DIQ", "SMQ")
  bp_mod <- ifelse(is_pre, "BPXO", "BPX")
  all_mods <- c(mods, bp_mod)

  dfs <- lapply(all_mods, function(m) {
    fname <- ifelse(is_pre, paste0("P_", m, ".xpt"), paste0(m, "_", suffix, ".xpt"))
    read_xpt(paste0(base, fname))
  })
  df <- Reduce(function(x, y) left_join(x, y, by = "SEQN"), dfs)

  if (is_pre) {
    df$sbp <- rowMeans(select(df, BPXOSY1, BPXOSY2, BPXOSY3), na.rm = TRUE)
    df$dbp <- rowMeans(select(df, BPXODI1, BPXODI2, BPXODI3), na.rm = TRUE)
    df$mec_wt <- df$WTMECPRP
  } else {
    df$sbp <- rowMeans(select(df, BPXSY1, BPXSY2, BPXSY3), na.rm = TRUE)
    df$dbp <- rowMeans(select(df, BPXDI1, BPXDI2, BPXDI3), na.rm = TRUE)
    df$mec_wt <- df$WTMEC2YR
  }
  df$cohort <- label
  return(df)
}

cat("Descargando 4 cohortes independientes (2011-2020)...\n")
c1 <- load_cycle("2011", "G", "NHANES 2011-2012")
c2 <- load_cycle("2013", "H", "NHANES 2013-2014")
c3 <- load_cycle("2015", "I", "NHANES 2015-2016")
c4 <- load_cycle("2017", "P", "NHANES 2017-2020", is_pre = TRUE)

raw_meta <- bind_rows(c1, c2, c3, c4)
cat("Descarga completa. N inicial combinado:", nrow(raw_meta), "\n")
```

```r
df_meta <- raw_meta %>%
  filter(RIDAGEYR >= 30 & RIDAGEYR <= 79) %>%
  filter(is.na(RIDEXPRG) | RIDEXPRG != 1) %>%
  mutate(
    age = as.numeric(RIDAGEYR),
    female = ifelse(RIAGENDR == 2, 1, 0),
    sex = ifelse(female == 1, "female", "male"),
    bmi = as.numeric(BMXBMI),
    waist = as.numeric(BMXWAIST),
    hba1c = as.numeric(LBXGH),
    fpg = as.numeric(LBXGLU),
    tc = as.numeric(LBXTC),
    hdl_c = as.numeric(LBDHDD),
    tg = as.numeric(LBXTR),
    scr = as.numeric(LBXSCR),
    uacr = (as.numeric(URXUMA) / as.numeric(URXUCR)) * 100,
    kappa = ifelse(female == 1, 0.7, 0.9),
    alpha = ifelse(female == 1, -0.241, -0.302),
    sex_mult = ifelse(female == 1, 1.012, 1.000),
    egfr = 142 * (pmin(scr / kappa, 1)^alpha) * (pmax(scr / kappa, 1)^(-1.200)) * (0.9938^age) * sex_mult,
    bp_med = ifelse(!is.na(BPQ050A) & BPQ050A == 1, 1, 0),
    statin = ifelse(!is.na(BPQ100D) & BPQ100D == 1, 1, 0),
    dm_dx = ifelse((!is.na(DIQ010) & DIQ010 == 1) | (!is.na(DIQ050) & DIQ050 == 1) |
                   (!is.na(DIQ070) & DIQ070 == 1) | (!is.na(hba1c) & hba1c >= 6.5), 1, 0),
    smoker = ifelse(!is.na(SMQ040) & SMQ040 %in% c(1, 2), 1, 0),
    cvd_hx = ifelse((!is.na(MCQ160B) & MCQ160B == 1) | (!is.na(MCQ160C) & MCQ160C == 1) |
                    (!is.na(MCQ160D) & MCQ160D == 1) | (!is.na(MCQ160E) & MCQ160E == 1) |
                    (!is.na(MCQ160F) & MCQ160F == 1), 1, 0),
    pir = as.numeric(INDFMPIR),
    ses_group = case_when(
      pir >= 4.00 ~ "1_High",
      pir >= 2.00 ~ "2_UpperMid",
      pir >= 1.00 ~ "3_LowerMid",
      pir < 1.00 ~ "4_LowPoverty",
      TRUE ~ NA_character_
    ),
    age_group = ifelse(age < 50, "Premature (30-49y)", "Older (50-79y)")
  ) %>%
  filter(!is.na(ses_group), !is.na(egfr), !is.na(uacr), !is.na(sbp),
         !is.na(bmi), !is.na(hba1c), !is.na(tc), !is.na(hdl_c),
         !is.na(mec_wt), mec_wt > 0)

age_c <- (df_meta$age - 55.0) / 10.0
non_hdl_c <- ((df_meta$tc - df_meta$hdl_c) * 0.02586) - 3.5
hdl_c <- ((df_meta$hdl_c * 0.02586) - 1.3) / 0.3
sbp_lt110 <- (pmin(df_meta$sbp, 110.0) - 110.0) / 20.0
sbp_ge110 <- (pmax(df_meta$sbp, 110.0) - 130.0) / 20.0
egfr_lt60 <- (pmin(df_meta$egfr, 60.0) - 60.0) / -15.0
egfr_ge60 <- (pmax(df_meta$egfr, 60.0) - 90.0) / -15.0

xb_f <- -3.307728 + 0.7939329 * age_c + 0.0305239 * non_hdl_c - 0.1606857 * hdl_c -
  0.2394003 * sbp_lt110 + 0.360078 * sbp_ge110 + 0.8667604 * df_meta$dm_dx +
  0.5360739 * df_meta$smoker + 0.6045917 * egfr_lt60 + 0.0433769 * egfr_ge60 +
  0.3151672 * df_meta$bp_med - 0.1477655 * df_meta$statin -
  0.0663612 * df_meta$bp_med * sbp_ge110 + 0.1197879 * df_meta$statin * non_hdl_c -
  0.0819715 * age_c * non_hdl_c + 0.0306769 * age_c * hdl_c -
  0.0946348 * age_c * sbp_ge110 - 0.27057 * age_c * df_meta$dm_dx -
  0.078715 * age_c * df_meta$smoker - 0.1637806 * age_c * egfr_lt60

xb_m <- -3.031168 + 0.7688528 * age_c + 0.0736174 * non_hdl_c - 0.0954431 * hdl_c -
  0.4347345 * sbp_lt110 + 0.3362658 * sbp_ge110 + 0.7692857 * df_meta$dm_dx +
  0.4386871 * df_meta$smoker + 0.5378979 * egfr_lt60 + 0.0164827 * egfr_ge60 +
  0.288879 * df_meta$bp_med - 0.1337349 * df_meta$statin -
  0.0475924 * df_meta$bp_med * sbp_ge110 + 0.150273 * df_meta$statin * non_hdl_c -
  0.0517874 * age_c * non_hdl_c + 0.0191169 * age_c * hdl_c -
  0.1049477 * age_c * sbp_ge110 - 0.2251948 * age_c * df_meta$dm_dx -
  0.0895067 * age_c * df_meta$smoker - 0.1543702 * age_c * egfr_lt60

df_meta$prevent_10yr <- (exp(ifelse(df_meta$female == 1, xb_f, xb_m)) / (1 + exp(ifelse(df_meta$female == 1, xb_f, xb_m)))) * 100

df_meta <- df_meta %>%
  mutate(
    ckd_very_high = (egfr < 30) | (egfr >= 30 & egfr < 45 & uacr >= 30) | (egfr >= 45 & egfr < 60 & uacr >= 300),
    ckd_mod_high = !ckd_very_high & ((egfr >= 30 & egfr < 60) | (egfr >= 60 & uacr >= 30)),
    has_htn = (sbp >= 130 | (!is.na(dbp) & dbp >= 80) | bp_med == 1),
    has_dm = (dm_dx == 1 | (!is.na(fpg) & fpg >= 126)),
    has_htg = (!is.na(tg) & tg >= 135),
    has_adiposity = (bmi >= 25) | (female == 1 & !is.na(waist) & waist >= 88) | (female == 0 & !is.na(waist) & waist >= 102),
    has_predm = (hba1c >= 5.7 & hba1c < 6.5) | (!is.na(fpg) & fpg >= 100 & fpg < 126),
    ckm_stage = case_when(
      cvd_hx == 1 ~ "Stage 4",
      prevent_10yr >= 20 | ckd_very_high ~ "Stage 3",
      has_htn | has_dm | has_htg | ckd_mod_high ~ "Stage 2",
      has_adiposity | has_predm ~ "Stage 1",
      TRUE ~ "Stage 0"
    ),
    adv_ckm = ifelse(ckm_stage %in% c("Stage 3", "Stage 4"), 1, 0)
  )

# Etapa 1: Estimacion ponderada independiente por cohorte
cohorts <- unique(df_meta$cohort)
stage1_res <- bind_rows(lapply(cohorts, function(ch) {
  sub <- filter(df_meta, cohort == ch)
  des <- svydesign(id = ~SDMVPSU, strata = ~SDMVSTRA, weights = ~mec_wt, nest = TRUE, data = sub)
  fit <- svyglm(adv_ckm ~ ses_group + age + sex + smoker, design = des, family = quasibinomial())
  s <- summary(fit)$coefficients
  terms <- c("ses_group2_UpperMid", "ses_group3_LowerMid", "ses_group4_LowPoverty")
  data.frame(
    cohort = ch,
    n_cohort = nrow(sub),
    term = terms,
    yi = s[terms, "Estimate"],
    sei = s[terms, "Std. Error"]
  )
}))

# Etapa 2: Metanalisis de efectos aleatorios (REML) con metafor
cat("\n=== RESULTADOS DEL METANALISIS DE EFECTOS ALEATORIOS (N TOTAL =", nrow(df_meta), ") ===\n")
meta_pooled <- bind_rows(lapply(unique(stage1_res$term), function(tm) {
  sub_m <- filter(stage1_res, term == tm)
  res_rma <- rma(yi = yi, sei = sei, data = sub_m, method = "REML")
  cat(sprintf("Contraste: %s | Pooled aOR = %.2f (95%% CI: %.2f - %.2f) | p = %.4g | I2 = %.1f%% (Q p = %.3f)\n",
              tm, exp(res_rma$beta[1]), exp(res_rma$ci.lb), exp(res_rma$ci.ub), res_rma$pval, res_rma$I2, res_rma$QEp))
  data.frame(
    cohort = "Pooled Random-Effects",
    n_cohort = sum(sub_m$n_cohort),
    term = tm,
    yi = as.numeric(res_rma$beta[1]),
    sei = as.numeric(res_rma$se),
    I2 = res_rma$I2,
    pval = res_rma$pval
  )
}))
```

```r
# Figura 1: Forest Plot de Metanalisis (4 cohortes + Diamante Pooled) para Pobreza vs Alto SES
fp_data <- bind_rows(
  filter(stage1_res, term == "ses_group4_LowPoverty") %>% mutate(type = "Cohort"),
  filter(meta_pooled, term == "ses_group4_LowPoverty") %>% mutate(type = "Pooled")
) %>%
  mutate(
    aOR = exp(yi),
    ci_low = exp(yi - 1.96 * sei),
    ci_high = exp(yi + 1.96 * sei),
    label_y = paste0(cohort, " (n=", format(n_cohort, big.mark = ","), ")"),
    label_y = factor(label_y, levels = rev(label_y)),
    or_text = sprintf("%.2f (%.2f–%.2f)", aOR, ci_low, ci_high)
  )

i2_val <- round(filter(meta_pooled, term == "ses_group4_LowPoverty")$I2, 1)

p_meta_forest <- ggplot(fp_data, aes(x = aOR, y = label_y, color = type, shape = type, size = type)) +
  geom_vline(xintercept = 1, linetype = "dashed", color = "gray45", linewidth = 0.8) +
  geom_errorbar(aes(xmin = ci_low, xmax = ci_high), width = 0.2, linewidth = 1) +
  geom_point() +
  geom_text(aes(label = or_text), vjust = -1.1, size = 3.8, fontface = "bold", show.legend = FALSE) +
  scale_x_log10(limits = c(0.85, 6.0), breaks = c(1.0, 1.5, 2.0, 3.0, 4.5)) +
  scale_color_manual(values = c("Cohort" = "#2171B5", "Pooled" = "#041E42")) +
  scale_shape_manual(values = c("Cohort" = 16, "Pooled" = 18)) +
  scale_size_manual(values = c("Cohort" = 3.8, "Pooled" = 6.0)) +
  labs(
    title = "A. Multi-Cohort Meta-Analysis: Poverty (PIR <1.0) vs. High SES on Advanced CKM",
    subtitle = paste0("Random-Effects Model (REML) | Total N = ", format(nrow(df_meta), big.mark = ","),
                      " | Heterogeneity I² = ", i2_val, "%"),
    x = "Adjusted Odds Ratio for Advanced CKM Stages 3–4 (95% CI, log scale)",
    y = NULL
  ) +
  theme_minimal(base_size = 13) +
  theme(plot.title = element_text(face = "bold", size = 13), legend.position = "none")

# Figura 2: Metanalisis de los 3 gradientes socioeconomicos agrupados (Dose-Response)
grad_data <- meta_pooled %>%
  mutate(
    aOR = exp(yi),
    ci_low = exp(yi - 1.96 * sei),
    ci_high = exp(yi + 1.96 * sei),
    Tier = case_when(
      term == "ses_group2_UpperMid" ~ "Upper-Middle (PIR 2.0–3.99)",
      term == "ses_group3_LowerMid" ~ "Lower-Middle (PIR 1.0–1.99)",
      term == "ses_group4_LowPoverty" ~ "Low / Poverty (PIR <1.0)"
    ),
    Tier = factor(Tier, levels = c("Low / Poverty (PIR <1.0)", "Lower-Middle (PIR 1.0–1.99)", "Upper-Middle (PIR 2.0–3.99)")),
    or_text = sprintf("%.2f (%.2f–%.2f)", aOR, ci_low, ci_high)
  )

p_meta_grad <- ggplot(grad_data, aes(x = aOR, y = Tier)) +
  geom_vline(xintercept = 1, linetype = "dashed", color = "gray45", linewidth = 0.8) +
  geom_errorbar(aes(xmin = ci_low, xmax = ci_high), width = 0.2, color = "#084594", linewidth = 1.1) +
  geom_point(size = 4.5, color = "#041E42") +
  geom_text(aes(label = or_text), vjust = -1.2, size = 3.8, fontface = "bold", color = "#041E42") +
  scale_x_log10(limits = c(0.9, 4.2), breaks = c(1.0, 1.5, 2.0, 2.5, 3.5)) +
  labs(
    title = "B. Pooled Dose-Response Gradient Across Socioeconomic Tiers (Ref: PIR 4.0+)",
    subtitle = "Adjusted for age, sex, and smoking across 4 independent national cohorts (2011–2020)",
    x = "Pooled Adjusted Odds Ratio (95% CI, log scale)",
    y = NULL
  ) +
  theme_minimal(base_size = 13) +
  theme(plot.title = element_text(face = "bold", size = 13))

ggsave("HMS_MetaAnalysis_Poster_BLUE.png", p_meta_forest / p_meta_grad, width = 12, height = 10, dpi = 300)
cat("Grafico de metanalisis guardado: HMS_MetaAnalysis_Poster_BLUE.png\n")
```

```r
library(splines)

df_meta <- df_meta %>%
  mutate(
    wt_pool = mec_wt / 4,
    strata_pool = paste0(cohort, "_", SDMVSTRA)
  )

ns_m <- ns(df_meta$age, df = 3)
df_meta$sp1 <- ns_m[, 1]
df_meta$sp2 <- ns_m[, 2]
df_meta$sp3 <- ns_m[, 3]

des_m <- svydesign(id = ~SDMVPSU, strata = ~strata_pool, weights = ~wt_pool, nest = TRUE, data = df_meta)
mod_m <- svyglm(adv_ckm ~ sp1 + sp2 + sp3 + ses_group + sex + smoker + cohort, design = des_m, family = quasibinomial())

ages <- 30:79
sp_g <- predict(ns_m, newx = ages)
ses_v <- c("1_High", "2_UpperMid", "3_LowerMid", "4_LowPoverty")

traj_m <- bind_rows(lapply(ses_v, function(s) {
  bind_rows(lapply(seq_along(ages), function(i) {
    tmp <- transform(df_meta, sp1 = sp_g[i, 1], sp2 = sp_g[i, 2], sp3 = sp_g[i, 3], ses_group = s)
    p <- as.numeric(predict(mod_m, newdata = tmp, type = "response"))
    data.frame(age = ages[i], ses_group = s, prob = weighted.mean(p, w = df_meta$wt_pool) * 100)
  }))
})) %>%
  mutate(SES_Tier = factor(recode(ses_group,
    "1_High" = "High SES (PIR 4.0+)",
    "2_UpperMid" = "Upper-Middle (PIR 2.0-3.99)",
    "3_LowerMid" = "Lower-Middle (PIR 1.0-1.99)",
    "4_LowPoverty" = "Low / Poverty (Under 1.0)"),
    levels = c("Low / Poverty (Under 1.0)", "Lower-Middle (PIR 1.0-1.99)",
               "Upper-Middle (PIR 2.0-3.99)", "High SES (PIR 4.0+)")))

p_obs_m <- weighted.mean(df_meta$adv_ckm, w = df_meta$wt_pool) * 100
a_hi <- traj_m %>% filter(ses_group == "1_High", prob >= p_obs_m) %>% slice(1) %>% pull(age)
a_lo <- traj_m %>% filter(ses_group == "4_LowPoverty", prob >= p_obs_m) %>% slice(1) %>% pull(age)
cat(sprintf("Prevalencia agrupada: %.1f%% | Pobreza la alcanza a los %d anos vs Alto SES a los %d anos (Brecha: %d anos)\n",
            p_obs_m, a_lo, a_hi, a_hi - a_lo))

p3_meta <- ggplot(traj_m, aes(x = age, y = prob, color = SES_Tier)) +
  geom_hline(yintercept = p_obs_m, linetype = "dotted", color = "gray45", linewidth = 0.8) +
  geom_line(linewidth = 1.4) +
  scale_color_manual(values = c("#041E42", "#084594", "#2171B5", "#6BAED6")) +
  scale_x_continuous(breaks = seq(30, 80, by = 10)) +
  scale_y_continuous(labels = function(x) paste0(round(x), "%")) +
  labs(
    title = "C. Pooled Age-Trajectory of Advanced CKM (2011-2020, N = 15,046)",
    subtitle = "Decade-pooled standardized probability (Stages 3-4) adjusted for sex, smoking & cohort",
    x = "Chronological Age (Years)",
    y = "Predicted Probability of Advanced CKM (%)",
    color = "Socioeconomic Tier"
  ) +
  theme_minimal(base_size = 13) +
  theme(plot.title = element_text(face = "bold", size = 13), legend.position = "bottom")

ggsave("HMS_MetaAnalysis_Fig3_Trajectory_BLUE.png", p3_meta, width = 10, height = 6, dpi = 300)
cat("OK - HMS_MetaAnalysis_Fig3_Trajectory_BLUE.png guardada\n")
```
